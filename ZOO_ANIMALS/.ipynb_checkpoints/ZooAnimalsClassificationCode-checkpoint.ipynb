```python
for i in l :
    for img in os.listdir(os.path.join(data,i)):
        imagePath = os.path.join(data,i, img)
        image = cv2.imread(imagePath)
        print(image)
```

```python
for i in l :
    for img in os.listdir(os.path.join(data,i)):
        imagePath = os.path.join(data,i, img)
        # image = cv2.imread(imagePath)
        tip = imghdr.what(imagePath)
        print(tip)
```

```python
for i in l :
    for img in os.listdir(os.path.join(data,i)):
        imagePath = os.path.join(data,i,img)
        tip = imghdr.what(imagePath)
        if tip not in image_exts:
            print('Image not in ext list{}'.format(imagePath))
            os.remove(imagePath)
```
