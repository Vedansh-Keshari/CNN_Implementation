```python
l = len(data)
```

```python
trainSize = round(l*0.70)
testSize = round(l*0.30) + 1
```

```python
trainSize , testSize
```

```python
train = data.take(trainSize)
test = data.skip(trainSize).take(testSize)
```

```python
len(train) , len(test)
```

```python
train
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Dropout
```

```python
model = Sequential()
```

```python
Conv2D??
```

```python
model.add(Conv2D(16, (3,3), 1, activation = 'relu', input_shape = (256,256,3)))
model.add(MaxPooling2D())

model.add(Conv2D(32, (3,3), 1, activation = 'relu'))
model.add(MaxPooling2D())

model.add(Conv2D(16, (3,3), 1, activation = 'relu'))
model.add(MaxPooling2D())

model.add(Flatten())

model.add(Dense(256, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
tf.optimizers??
```

```python
model.compile('adam', loss=tf.losses.BinaryCrossentropy(), metrics=['accuracy'])
```

```python
hist = model.fit(train, epochs = 20)
```

```python
fig = plt.figure()
plt.plot(hist.history['loss'], color='teal')
plt.xlabel('No of Epochs')
plt.ylabel('Loss')
plt.grid()
plt.show()
```

```python
fig = plt.figure()
plt.plot(hist.history['accuracy'], color='teal')
plt.xlabel('No of Epochs')
plt.ylabel('Accuracy')
plt.grid()
plt.show()
```
